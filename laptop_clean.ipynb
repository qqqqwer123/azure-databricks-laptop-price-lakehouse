```python
import pandas as pd

file_name = r'C:\Users\USER\Desktop\a找工作\作品集\預測模型\資料集\amazon_laptop_prices_v01.csv'
df=pd.read_csv(file_name)


#查看數據型態
print(df.info())

# 強制顯示所有欄位 (不隱藏中間欄位)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

df.head()
```

```python
#df['price']=df['price'].astype(str).str.replace('$','').str.replace(',','')

#先把貨幣符號提取出來
df['currency_symbol']=df['price'].astype(str).str.extract('([\$€£¥])')
# 看看你的符號欄位裡到底有哪些不同的東西
print(df['currency_symbol'].value_counts())
#只有美元符號，那可以全部移除掉
df['price']=df['price'].astype(str).str.replace('$','').str.replace(',','')
df['price']=pd.to_numeric(df['price'],errors='coerce')
df.head()
```

```python
#處理RAM
df['ram'].value_counts() #沒有TB，可以直接處理掉
df=df.rename(columns={'ram': 'ram(GB)'})
df['ram(GB)']=df['ram(GB)'].astype(str).str.extract('(\d+)')
```

```python
#處理harddist
df['harddisk'].value_counts() #沒有TB，可以直接處理掉
#1TB=1024GB,1GB=1024MB,1MB=1024KB
#def clean_harddisk(value): 常用的話之後再包成function
df['temp_storage']=df['harddisk'].str.extract('(\d+)').astype(float)
df['temp_storage_2']=df['harddisk'].str.contains('TB',case=False).map({True: 1024, False: 1}).astype(float)
df['harddisk_new_GB']=df['temp_storage']*df['temp_storage_2']
#TEMP可以刪了
df=df.drop(columns=['temp_storage','temp_storage_2'])
```

```python
#處理screen_size
df['screen_size'].value_counts()
df['screen_size']=df['screen_size'].astype('str').str.extract('(\d+)')
df['screen_size']=pd.to_numeric(df['screen_size'],errors='coerce')
```

```python
#檢查missing value
#看一下harddisk是null的資料
missing_value=df[df['harddisk'].isnull()]
print(missing_value)

#看起來有很多NULL 看能不能用相同型號的harddisk補
#所以要先檢查找出同型號中，硬碟容量有幾種不同的值，如果都相同就可以直接補
model_check = df.groupby('model')['harddisk_new_GB'].nunique()
print("同一型號擁有多種硬碟容量的數量：")
print(model_check[model_check > 1].count()) #不行 用填補的

#那只好把price和harddisk、screen_size是null的都刪除
df=df.dropna(subset=['price','harddisk_new_GB','screen_size'])
```

```python
#在進行資料拆分的時候發現有些品牌很小、或者是同品牌大小寫不一致，回來補做清理
#少於一筆的資料拆分會讓上面的程式報錯
#Step 1：統一品牌名稱（先做這個）
df['brand'] = df['brand'].str.lower()
#Step 2：刪掉太少的品牌（關鍵），我選擇刪掉不到三筆的
counts = df['brand'].value_counts()
valid_brands = counts[counts >= 3].index
df = df[df['brand'].isin(valid_brands)]
```

```python
# 儲存清理好的資料，index=False 是為了不要多出一列沒用的索引序號
df.to_csv('cleaned_laptop_data.csv', index=False, encoding='utf-8-sig')

print("資料清洗完成，已儲存至 cleaned_laptop_data.csv")
```
